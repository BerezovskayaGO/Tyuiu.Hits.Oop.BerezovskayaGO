<h2 style="color:DodgerBlue">Описание проекта:</h2>

----
Описание задачи: 
Создать базовый класс Employee в C#, который будет представлять информацию о 
сотрудниках компании. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 



Требования к базовому классу Employee: 

• Атрибуты: Имя (Name), Возраст (Age), Зарплата (Salary).

• Методы: 
    1. GetInfo(): метод для получения информации о сотруднике в виде 
строки. 
    2. Work(): метод для вывода сообщения о выполнении рабочих 
обязанностей. 
    3. TakeVacation(): метод для вывода сообщения о взятии отпуска. 




Требования к производным классам: 
1. Менеджер (Manager): Должен содержать дополнительные атрибуты, такие 
как Количество подчиненных (SubordinatesCount). Метод Work() должен быть 
переопределен для добавления информации о управлении командой. 
2. Разработчик (Developer): Должен содержать дополнительные атрибуты, 
такие как Специализация (Specialization). Метод TakeVacation() должен быть 
переопределен для добавления информации о необходимости согласования 
отпуска с проектными сроками. 
3. Дизайнер (Designer) (если требуется третий класс): Должен содержать 
дополнительные атрибуты, такие как Стиль дизайна (DesignStyle). 
Метод Work() должен быть переопределен для добавления информации о 
работе над дизайн-проектами.


#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями 


<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

// Базовый интерфейс для всех сотрудников
public interface IEmployee
{
    string GetInfo();
    void Work();
    void TakeVacation();
    string Name { get; }
    int Age { get; }
    decimal Salary { get; }
}

// Базовый класс Employee
public class Employee : IEmployee
{
    public string Name { get; set; }
    public int Age { get; set; }
    public decimal Salary { get; set; }

    public Employee(string name, int age, decimal salary)
    {
        Name = name;
        Age = age;
        Salary = salary;
    }

    public virtual string GetInfo()
    {
        return $"Сотрудник: {Name}, Возраст: {Age}, Зарплата: {Salary} руб.";
    }

    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет рабочие обязанности");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} берет отпуск");
    }
}

// Класс Manager с дополнительными атрибутами и переопределенными методами
public class Manager : Employee
{
    public int SubordinatesCount { get; set; }
    public string Department { get; set; }
    public int YearsAsManager { get; set; }
    public decimal Bonus { get; set; }

    public Manager(string name, int age, decimal salary, int subordinatesCount, string department) 
        : base(name, age, salary)
    {
        SubordinatesCount = subordinatesCount;
        Department = department;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Должность: Менеджер, Подчиненных: {SubordinatesCount}, Отдел: {Department}";
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} управляет командой из {SubordinatesCount} человек в отделе {Department}");
    }

    // Новые методы для менеджера
    public void ConductMeeting()
    {
        Console.WriteLine($"{Name} проводит совещание с командой");
    }

    public void ReviewPerformance()
    {
        Console.WriteLine($"{Name} проводит оценку эффективности сотрудников");
    }
}

// Класс Developer с дополнительными атрибутами и переопределенными методами
public class Developer : Employee
{
    public string Specialization { get; set; }
    public string ProgrammingLanguage { get; set; }
    public int CompletedProjects { get; set; }
    public bool RemoteWork { get; set; }

    public Developer(string name, int age, decimal salary, string specialization, string programmingLanguage) 
        : base(name, age, salary)
    {
        Specialization = specialization;
        ProgrammingLanguage = programmingLanguage;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Должность: Разработчик, Специализация: {Specialization}, Язык: {ProgrammingLanguage}";
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"{Name} берет отпуск. Необходимо согласование с проектными сроками!");
    }

    // Новые методы для разработчика
    public void WriteCode()
    {
        Console.WriteLine($"{Name} пишет код на {ProgrammingLanguage}");
    }

    public void DebugCode()
    {
        Console.WriteLine($"{Name} отлаживает код");
    }
}

// Класс Designer с дополнительными атрибутами и переопределенными методами
public class Designer : Employee
{
    public string DesignStyle { get; set; }
    public string DesignTool { get; set; }
    public string Portfolio { get; set; }
    public int DesignAwards { get; set; }

    public Designer(string name, int age, decimal salary, string designStyle, string designTool) 
        : base(name, age, salary)
    {
        DesignStyle = designStyle;
        DesignTool = designTool;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Должность: Дизайнер, Стиль: {DesignStyle}, Инструмент: {DesignTool}";
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} работает над дизайн-проектами в стиле {DesignStyle} используя {DesignTool}");
    }

    // Новые методы для дизайнера
    public void CreatePrototype()
    {
        Console.WriteLine($"{Name} создает прототип дизайна");
    }

    public void PresentDesign()
    {
        Console.WriteLine($"{Name} представляет дизайн клиенту");
    }
}

// Интерфейс для сотрудников, работающих над проектами
public interface IProjectMember
{
    void JoinProject(string projectName);
    void LeaveProject(string projectName);
    string GetProjectStatus();
}

// Явная реализация интерфейса IProjectMember для Manager
public class ProjectManager : Manager, IProjectMember
{
    private List<string> currentProjects = new List<string>();
    
    public ProjectManager(string name, int age, decimal salary, int subordinatesCount, string department) 
        : base(name, age, salary, subordinatesCount, department) { }

    // Явная реализация методов интерфейса
    void IProjectMember.JoinProject(string projectName)
    {
        currentProjects.Add(projectName);
        Console.WriteLine($"{Name} присоединился к проекту '{projectName}' как менеджер проекта");
    }

    void IProjectMember.LeaveProject(string projectName)
    {
        if (currentProjects.Remove(projectName))
            Console.WriteLine($"{Name} покинул проект '{projectName}'");
    }

    string IProjectMember.GetProjectStatus()
    {
        return $"{Name} управляет {currentProjects.Count} проектами: {string.Join(", ", currentProjects)}";
    }
}

// Generic класс для управления сотрудниками
public class EmployeeManager<T> where T : Employee
{
    private List<T> employees = new List<T>();
    
    public void AddEmployee(T employee)
    {
        employees.Add(employee);
        Console.WriteLine($"Сотрудник {employee.Name} добавлен в менеджер");
    }
    
    public void RemoveEmployee(T employee)
    {
        if (employees.Remove(employee))
            Console.WriteLine($"Сотрудник {employee.Name} удален из менеджера");
    }
    
    public void DisplayAllEmployees()
    {
        Console.WriteLine($"\n--- Все сотрудники в менеджере ({typeof(T).Name}) ---");
        foreach (var emp in employees)
        {
            Console.WriteLine(emp.GetInfo());
        }
    }

    public void MakeAllWork()
    {
        Console.WriteLine($"\n--- Работа всех сотрудников ({typeof(T).Name}) ---");
        foreach (var emp in employees)
        {
            emp.Work();
        }
    }
}

Manager manager = new Manager("Иван Иванов", 35, 80000, 5, "IT");
Developer developer = new Developer("Анна Петрова", 28, 70000, "Backend", "C#");
Designer designer = new Designer("Мария Попова", 25, 65000, "Минимализм", "Figma");
ProjectManager projectManager = new ProjectManager("Сергей Сидоров", 40, 90000, 8, "Разработка");

Employee[] employees = { manager, developer, designer, projectManager };

Console.WriteLine("ИНФОРМАЦИЯ О ВСЕХ СОТРУДНИКАХ:");
foreach (var emp in employees)
{
    Console.WriteLine(emp.GetInfo());
}
Console.WriteLine();

Console.WriteLine("РАБОЧИЕ ОБЯЗАННОСТИ (полиморфизм):");
foreach (var emp in employees)
{
    emp.Work();
}
Console.WriteLine();

Console.WriteLine("ОТПУСКА (полиморфизм):");
foreach (var emp in employees)
{
    emp.TakeVacation();
}
Console.WriteLine();

Console.WriteLine("УНИКАЛЬНЫЕ МЕТОДЫ:");
manager.ConductMeeting();
developer.WriteCode();
designer.CreatePrototype();
Console.WriteLine();

Console.WriteLine("РАБОТА С ИНТЕРФЕЙСОМ IProjectMember:");
IProjectMember projectMember = projectManager;
projectMember.JoinProject("Мобильное приложение");
projectMember.JoinProject("Веб-платформа");
Console.WriteLine(projectMember.GetProjectStatus());
Console.WriteLine();

Console.WriteLine("РАБОТА С GENERIC КЛАССАМИ:");
        
EmployeeManager<Developer> devManager = new EmployeeManager<Developer>();
devManager.AddEmployee(developer);
devManager.DisplayAllEmployees();
devManager.MakeAllWork();

EmployeeManager<Employee> allManager = new EmployeeManager<Employee>();
foreach (var emp in employees)
{
    allManager.AddEmployee(emp);
}
allManager.DisplayAllEmployees();

Console.WriteLine("\n=== ПРОГРАММА ЗАВЕРШЕНА ===");
```
