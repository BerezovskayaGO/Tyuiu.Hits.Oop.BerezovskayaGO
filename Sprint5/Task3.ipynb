Описание задачи: 
Создать базовый класс Employee в C#, который будет представлять информацию о 
сотрудниках компании. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 



Требования к базовому классу Employee: 

• Атрибуты: Имя (Name), Возраст (Age), Зарплата (Salary).

• Методы: 
    1. GetInfo(): метод для получения информации о сотруднике в виде 
строки. 
    2. Work(): метод для вывода сообщения о выполнении рабочих 
обязанностей. 
    3. TakeVacation(): метод для вывода сообщения о взятии отпуска. 




Требования к производным классам: 
1. Менеджер (Manager): Должен содержать дополнительные атрибуты, такие 
как Количество подчиненных (SubordinatesCount). Метод Work() должен быть 
переопределен для добавления информации о управлении командой. 
2. Разработчик (Developer): Должен содержать дополнительные атрибуты, 
такие как Специализация (Specialization). Метод TakeVacation() должен быть 
переопределен для добавления информации о необходимости согласования 
отпуска с проектными сроками. 
3. Дизайнер (Designer) (если требуется третий класс): Должен содержать 
дополнительные атрибуты, такие как Стиль дизайна (DesignStyle). 
Метод Work() должен быть переопределен для добавления информации о 
работе над дизайн-проектами.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.

```
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void EmployeeEventHandler(string message);
public delegate void WorkCompletedHandler(string employeeName, string task);

public interface IEmployee
{
    string GetInfo();
    void Work();
    void TakeVacation();
    string Name { get; }
    int Age { get; }
    decimal Salary { get; }
}

public class Employee : IEmployee
{
    // Дополнительные атрибуты
    public string Email { get; set; }
    public DateTime HireDate { get; set; }
    public int VacationDays { get; set; }
    
    public event EmployeeEventHandler EmployeeEvent;
    public event WorkCompletedHandler WorkCompleted;

    public string Name { get; set; }
    public int Age { get; set; }
    public decimal Salary { get; set; }

    public Employee(string name, int age, decimal salary, string email, DateTime hireDate)
    {
        Name = name;
        Age = age;
        Salary = salary;
        Email = email;
        HireDate = hireDate;
        VacationDays = 28;
    }

    protected virtual void OnEmployeeEvent(string message)
    {
        EmployeeEvent?.Invoke(message);
    }

    protected virtual void OnWorkCompleted(string task)
    {
        WorkCompleted?.Invoke(Name, task);
    }

    public virtual string GetInfo()
    {
        return $"Сотрудник: {Name}, Возраст: {Age}, Зарплата: {Salary} руб.";
    }

    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет рабочие обязанности");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} берет отпуск");
    }

    // Дополнительные методы
    public virtual void SendEmail(string recipient, string subject)
    {
        Console.WriteLine($"{Name} отправляет email: {subject}");
        OnWorkCompleted($"Отправка email: {subject}");
    }

    public virtual void AttendMeeting(string meetingTopic)
    {
        Console.WriteLine($"{Name} участвует в совещании: {meetingTopic}");
    }

    public virtual void UpdateProfile()
    {
        Console.WriteLine($"{Name} обновил профиль");
        OnEmployeeEvent($"{Name} обновил профиль");
    }
}

public class Manager : Employee
{
    // Дополнительные атрибуты
    public List<string> ManagedProjects { get; set; }
    public decimal Budget { get; set; }
    public int TeamSize { get; set; }

    public int SubordinatesCount { get; set; }
    public string Department { get; set; }
    public int YearsAsManager { get; set; }
    public decimal Bonus { get; set; }

    public Manager(string name, int age, decimal salary, int subordinatesCount, string department) 
        : base(name, age, salary, $"{name.ToLower()}@company.com", DateTime.Now.AddYears(-2))
    {
        SubordinatesCount = subordinatesCount;
        Department = department;
        ManagedProjects = new List<string>();
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Должность: Менеджер, Подчиненных: {SubordinatesCount}, Отдел: {Department}";
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} управляет командой из {SubordinatesCount} человек в отделе {Department}");
    }

    // Новые методы для менеджера
    public void ConductMeeting()
    {
        Console.WriteLine($"{Name} проводит совещание с командой");
    }

    public void ReviewPerformance()
    {
        Console.WriteLine($"{Name} проводит оценку эффективности сотрудников");
    }

    public override void SendEmail(string recipient, string subject)
    {
        Console.WriteLine($"{Name} (менеджер) отправляет важный email: {subject}");
        OnWorkCompleted($"Важный email: {subject}");
    }

    public override void UpdateProfile()
    {
        base.UpdateProfile();
        Console.WriteLine($"{Name} также обновил контакты для подчиненных");
    }
}

public class Developer : Employee
{
    // Дополнительные атрибуты
    public List<string> Technologies { get; set; }
    public int CodeCommits { get; set; }
    public string GitHub { get; set; }

    public string Specialization { get; set; }
    public string ProgrammingLanguage { get; set; }
    public int CompletedProjects { get; set; }
    public bool RemoteWork { get; set; }

    public Developer(string name, int age, decimal salary, string specialization, string programmingLanguage) 
        : base(name, age, salary, $"{name.ToLower()}@company.com", DateTime.Now.AddYears(-1))
    {
        Specialization = specialization;
        ProgrammingLanguage = programmingLanguage;
        Technologies = new List<string>();
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Должность: Разработчик, Специализация: {Specialization}, Язык: {ProgrammingLanguage}";
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"{Name} берет отпуск. Необходимо согласование с проектными сроками!");
    }

    // Новые методы для разработчика
    public void WriteCode()
    {
        Console.WriteLine($"{Name} пишет код на {ProgrammingLanguage}");
    }

    public void DebugCode()
    {
        Console.WriteLine($"{Name} отлаживает код");
    }

    public override void SendEmail(string recipient, string subject)
    {
        Console.WriteLine($"{Name} (разработчик) отправляет технический email: {subject}");
        OnWorkCompleted($"Технический email: {subject}");
    }

    public void AddTechnology(string technology)
    {
        Technologies.Add(technology);
        Console.WriteLine($"{Name} изучил новую технологию: {technology}");
        OnEmployeeEvent($"{Name} освоил {technology}");
    }
}

public class Designer : Employee
{
    // Дополнительные атрибуты
    public List<string> DesignTools { get; set; }
    public int DesignsCreated { get; set; }
    public string Behance { get; set; }

    public string DesignStyle { get; set; }
    public string DesignTool { get; set; }
    public string Portfolio { get; set; }
    public int DesignAwards { get; set; }

    public Designer(string name, int age, decimal salary, string designStyle, string designTool) 
        : base(name, age, salary, $"{name.ToLower()}@company.com", DateTime.Now.AddYears(-1))
    {
        DesignStyle = designStyle;
        DesignTool = designTool;
        DesignTools = new List<string> { designTool };
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Должность: Дизайнер, Стиль: {DesignStyle}, Инструмент: {DesignTool}";
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} работает над дизайн-проектами в стиле {DesignStyle} используя {DesignTool}");
    }

    // Новые методы для дизайнера
    public void CreatePrototype()
    {
        Console.WriteLine($"{Name} создает прототип дизайна");
    }

    public void PresentDesign()
    {
        Console.WriteLine($"{Name} представляет дизайн клиенту");
    }

    public override void SendEmail(string recipient, string subject)
    {
        Console.WriteLine($"{Name} (дизайнер) отправляет креативный email: {subject}");
        OnWorkCompleted($"Креативный email: {subject}");
    }

    public void AddDesignTool(string tool)
    {
        DesignTools.Add(tool);
        Console.WriteLine($"{Name} освоил новый инструмент: {tool}");
        OnEmployeeEvent($"{Name} освоил {tool}");
    }
}

// Коллекция для управления сотрудниками с событиями
public class Company
{
    private List<Employee> _employees = new List<Employee>();
    
    public event EmployeeEventHandler CompanyEvent;

    public void AddEmployee(Employee employee)
    {
        _employees.Add(employee);
        employee.EmployeeEvent += OnEmployeeActivity;
        employee.WorkCompleted += OnWorkCompleted;
        CompanyEvent?.Invoke($"Добавлен сотрудник: {employee.Name}");
    }

    private void OnEmployeeActivity(string message)
    {
        Console.WriteLine($"Активность: {message}");
    }

    private void OnWorkCompleted(string name, string task)
    {
        Console.WriteLine($"Завершено: {name} - {task}");
    }

    public void ShowAllEmployees()
    {
        Console.WriteLine("\nВсе сотрудники компании:");
        foreach (var emp in _employees)
        {
            Console.WriteLine(emp.GetInfo());
        }
    }
}

Manager manager = new Manager("Иван Иванов", 35, 80000, 5, "IT");
Developer developer = new Developer("Анна Петрова", 28, 70000, "Backend", "C#");
Designer designer = new Designer("Мария Попова", 25, 65000, "Минимализм", "Figma");

Company company = new Company();
company.CompanyEvent += (message) => Console.WriteLine($"Событие компании: {message}");

company.AddEmployee(manager);
company.AddEmployee(developer);
company.AddEmployee(designer);

Console.WriteLine("ИНФОРМАЦИЯ О ВСЕХ СОТРУДНИКАХ:");
foreach (var emp in new Employee[] { manager, developer, designer })
{
    Console.WriteLine(emp.GetInfo());
}

Console.WriteLine("\nРАБОЧИЕ ОБЯЗАННОСТИ:");
foreach (var emp in new Employee[] { manager, developer, designer })
{
    emp.Work();
}

Console.WriteLine("\nОТПУСКА:");
foreach (var emp in new Employee[] { manager, developer, designer })
{
    emp.TakeVacation();
}

Console.WriteLine("\nУНИКАЛЬНЫЕ МЕТОДЫ:");
manager.ConductMeeting();
developer.WriteCode();
designer.CreatePrototype();

Console.WriteLine("\nДОПОЛНИТЕЛЬНЫЕ МЕТОДЫ:");
manager.SendEmail("team@company.com", "План на неделю");
developer.SendEmail("techlead@company.com", "Вопрос по архитектуре");
designer.SendEmail("client@company.com", "Концепция дизайна");

Console.WriteLine("\nОБНОВЛЕНИЕ ПРОФИЛЕЙ:");
manager.UpdateProfile();
developer.AddTechnology("Docker");
designer.AddDesignTool("Adobe XD");

company.ShowAllEmployees();
```
