```polyglot-notebook
using System;
using System.Collections.Generic;

public abstract class Component
{
    private decimal _price;

    public decimal Price
    {
        get { return _price; }
        set 
        { 
            if (value >= 0)
                _price = value;
            else
                throw new ArgumentException("Цена не может быть отрицательной");
        }
    }

    public virtual string GetInfo()
    {
        return ($"Компонент стоимостью {Price} руб.");
    }

    public override string ToString()
    {
        return GetInfo();
    }
}

public interface IInstallable
{
    void Install();
}

public class CPU : Component, IInstallable  
{
    public override string GetInfo()
    {
        return ($"CPU стоимостью {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine("CPU установлен в сокет материнской платы");
    }
}

public class RAM : Component, IInstallable  
{
    public override string GetInfo()
    {
        return ($"RAM стоимостью {Price} руб.");
    }
    
    public void Install()
    {
        Console.WriteLine("RAM установлена в слоты памяти");
    }
}

public class GPU : Component, IInstallable  
{
    public override string GetInfo()
    {
        return ($"GPU стоимостью {Price} руб.");
    }
    
    public void Install()
    {
        Console.WriteLine("GPU установлена в PCI-Express слот");
    }
}

public class HDD : Component, IInstallable  
{
    public override string GetInfo()
    {
        return ($"HDD стоимостью {Price} руб.");
    }
    
    public void Install()
    {
        Console.WriteLine("HDD подключен к SATA разъему");
    }
}

public class Computer
{
    private List<Component> _components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        component.Install();
        Console.WriteLine($"Компонент добавлен: {component.GetInfo()}");
    }
    
    public void RemoveComponent<T>(T component) where T : Component
    {
        if (_components.Remove(component))
        {
            Console.WriteLine($"Компонент удален: {component.GetInfo()}");
        }
        else
        {
            Console.WriteLine("Компонент не найден");
        }
    }
    
    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (var component in _components)
        {
            total += component.Price;
        }
        return total;
    }
    
    public void DisplayComponents()
    {
        Console.WriteLine("\n--- Компоненты компьютера ---");
        foreach (var component in _components)
        {
            Console.WriteLine(component.GetInfo()); 
        }
        Console.WriteLine($"Общая стоимость: {GetTotalPrice()} руб.");
    }
}

public static class ComputerFactory
{
    public static Computer CreateComputerWithComponent<T>() where T : Component, IInstallable, new()
    {
        Computer computer = new Computer();
        T component = new T();
        component.Price = 10000; 
        computer.AddComponent(component);
        return computer;
    }
    
    public static Computer CreateGamingComputer()
    {
        Computer computer = new Computer();
        
        CPU cpu = new CPU { Price = 25000 };
        RAM ram = new RAM { Price = 8000 };
        GPU gpu = new GPU { Price = 50000 };
        HDD hdd = new HDD { Price = 5000 };
        
        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        computer.AddComponent(gpu);
        computer.AddComponent(hdd);
        
        return computer;
    }
}

Computer myComputer = new Computer();

CPU intelCPU = new CPU { Price = 20000 };
RAM corsairRAM = new RAM { Price = 6000 };
GPU nvidiaGPU = new GPU { Price = 45000 };
HDD seagateHDD = new HDD { Price = 4000 };

Console.WriteLine("Добавление компонентов:");
myComputer.AddComponent(intelCPU);
myComputer.AddComponent(corsairRAM);
myComputer.AddComponent(nvidiaGPU);
myComputer.AddComponent(seagateHDD);

myComputer.DisplayComponents();

Console.WriteLine("\nУдаление компонента:");
myComputer.RemoveComponent(seagateHDD);
myComputer.DisplayComponents();

Computer factoryComputer = ComputerFactory.CreateComputerWithComponent<CPU>();
factoryComputer.DisplayComponents();

Console.WriteLine("\nСоздание игрового компьютера:");
Computer gamingComputer = ComputerFactory.CreateGamingComputer();
gamingComputer.DisplayComponents();

Console.WriteLine("\n=== Демонстрация перегрузки ===");
Console.WriteLine($"CPU как строка: {intelCPU.ToString()}");
```
