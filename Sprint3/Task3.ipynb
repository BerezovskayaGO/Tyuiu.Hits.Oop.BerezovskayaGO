Описание задачи: 
Создать базовый класс Employee в C#, который будет представлять информацию о 
сотрудниках компании. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 



Требования к базовому классу Employee: 

• Атрибуты: Имя (Name), Возраст (Age), Зарплата (Salary).

• Методы: 
    1. GetInfo(): метод для получения информации о сотруднике в виде 
строки. 
    2. Work(): метод для вывода сообщения о выполнении рабочих 
обязанностей. 
    3. TakeVacation(): метод для вывода сообщения о взятии отпуска. 




Требования к производным классам: 
1. Менеджер (Manager): Должен содержать дополнительные атрибуты, такие 
как Количество подчиненных (SubordinatesCount). Метод Work() должен быть 
переопределен для добавления информации о управлении командой. 
2. Разработчик (Developer): Должен содержать дополнительные атрибуты, 
такие как Специализация (Specialization). Метод TakeVacation() должен быть 
переопределен для добавления информации о необходимости согласования 
отпуска с проектными сроками. 
3. Дизайнер (Designer) (если требуется третий класс): Должен содержать 
дополнительные атрибуты, такие как Стиль дизайна (DesignStyle). 
Метод Work() должен быть переопределен для добавления информации о 
работе над дизайн-проектами.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;

public class EmployeeManager<T> where T : Employee
{
    private List<T> employees = new List<T>();
    
    public void AddEmployee(T employee)
    {
        employees.Add(employee);
        Console.WriteLine($"Сотрудник {employee.Name} добавлен в менеджер");
    }
    
    public void RemoveEmployee(T employee)
    {
        if (employees.Remove(employee))
            Console.WriteLine($"Сотрудник {employee.Name} удален из менеджера");
    }
    
    public void DisplayAllEmployees()
    {
        Console.WriteLine($"\n--- Все сотрудники в менеджере ({typeof(T).Name}) ---");
        foreach (var emp in employees)
        {
            Console.WriteLine(emp.GetInfo());
        }
    }

    public void DisplayAllEmployees(bool detailed)
    {
        if (detailed)
        {
            Console.WriteLine($"\n--- Детальная информация о сотрудниках ({typeof(T).Name}) ---");
            foreach (var emp in employees)
            {
                Console.WriteLine(emp.GetDetailedInfo());
            }
        }
        else
        {
            DisplayAllEmployees();
        }
    }
}

public class Employee
{
    private string name;
    private int age;
    private decimal salary;
    private string email;
    private string education;
    private int experience;

    private string phoneNumber;
    private string department;

    public string Name
    {
        get { return name; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                name = value;
            else
                Console.WriteLine("Имя не может быть пустым!");
        }
    }

    public int Age
    {
        get { return age; }
        set 
        { 
            if (value >= 18 && value <= 65)
                age = value;
            else
                Console.WriteLine("Возраст должен быть от 18 до 65 лет!");
        }
    }

    public decimal Salary
    {
        get { return salary; }
        set 
        { 
            if (value >= 0)
                salary = value;
            else
                Console.WriteLine("Зарплата не может быть отрицательной!");
        }
    }

    public string Email
    {
        get { return email; }
        set 
        { 
            if (!string.IsNullOrEmpty(value) && value.Contains("@"))
                email = value;
            else
                Console.WriteLine("Email должен содержать @!");
        }
    }
    
    public string Education
    {
        get { return education; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                education = value;
            else
                Console.WriteLine("Образование не может быть пустым!");
        }
    }

    public int Experience
    {
        get { return experience; }
        set 
        { 
            if (value >= 0)
                experience = value;
            else
                Console.WriteLine("Опыт не должен быть отрицательным!");
        }
    }

    public string PhoneNumber
    {
        get { return phoneNumber; }
        set 
        { 
            if (!string.IsNullOrEmpty(value) && value.Length >= 10)
                phoneNumber = value;
            else
                Console.WriteLine("Номер телефона должен содержать не менее 10 символов!");
        }
    }
    
    public string Department
    {
        get { return department; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                department = value;
            else
                Console.WriteLine("Отдел не может быть пустым!");
        }
    }

    public Employee(string name, int age, decimal salary, string email, string education, int experience, string phoneNumber, string department)
    {
        Name = name;
        Age = age;
        Salary = salary;
        Email = email;
        Education = education;
        Experience = experience;
        PhoneNumber = phoneNumber;
        Department = department;
    }

    public virtual string GetInfo()
    {
        return $"Сотрудник: {Name}, Возраст: {Age}, Зарплата: {Salary} руб., Отдел: {Department}";
    }

    public virtual string GetDetailedInfo()
    {
        return $"Сотрудник: {Name}, Возраст: {Age}, Зарплата: {Salary} руб., Почта: {Email}, Образование: {Education}, Опыт: {Experience} лет, Телефон: {PhoneNumber}, Отдел: {Department}";
    }
    
    public virtual void UpdateContactInfo(string newEmail, string newPhone)
    {
        Email = newEmail;
        PhoneNumber = newPhone;
        Console.WriteLine($"{Name}: Контактная информация обновлена");
    }

    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет рабочие задачи в отделе {Department}");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} берет отпуск");
    }

    public virtual void TakeVacation(int days)
    {
        Console.WriteLine($"{Name} берет отпуск на {days} дней");
    }

    public virtual void InteractWith(Employee otherEmployee)
    {
        Console.WriteLine($"{Name} взаимодействует с {otherEmployee.Name}");
    }

    public virtual void InteractWith(Employee otherEmployee, string purpose)
    {
        Console.WriteLine($"{Name} взаимодействует с {otherEmployee.Name} для {purpose}");
    }
}

public interface IDesigner
{
    void CreatePrototype();
}

public interface IProjectMember
{
    void JoinProject(string projectName);
    void LeaveProject(string projectName);
}

public class Manager : Employee, IProjectMember
{
    private int subordinatesCount;
    private List<string> currentProjects = new List<string>();

    public int SubordinatesCount
    {
        get { return subordinatesCount; }
        set 
        { 
            if (value >= 0)
                subordinatesCount = value;
            else
                Console.WriteLine("Количество подчиненных не может быть отрицательным!");
        }
    }

    public Manager(string name, int age, decimal salary, string email, string education, int experience, string phoneNumber, string department, int subordinatesCount) 
        : base(name, age, salary, email, education, experience, phoneNumber, department)
    {
        SubordinatesCount = subordinatesCount;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Подчиненных: {SubordinatesCount}";
    }
    
    public override string GetDetailedInfo()
    {
        return base.GetDetailedInfo() + $", Подчиненных: {SubordinatesCount}, Проектов: {currentProjects.Count}";
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} управляет командой из {SubordinatesCount} человек в отделе {Department}");
    }

    public override void UpdateContactInfo(string newEmail, string newPhone)
    {
        base.UpdateContactInfo(newEmail, newPhone);
        Console.WriteLine($"{Name}: Уведомление отправлено всем {SubordinatesCount} подчиненным");
    }

    public override void InteractWith(Employee otherEmployee)
    {
        if (otherEmployee is Developer)
            Console.WriteLine($"{Name} ставит задачи разработчику {otherEmployee.Name}");
        else if (otherEmployee is Designer)
            Console.WriteLine($"{Name} обсуждает требования с дизайнером {otherEmployee.Name}");
        else
            Console.WriteLine($"{Name} проводит встречу с {otherEmployee.Name}");
    }

    public override void InteractWith(Employee otherEmployee, string purpose)
    {
        Console.WriteLine($"{Name} проводит встречу с {otherEmployee.Name} по поводу: {purpose}");
    }

    public void ConductMeeting()
    {
        Console.WriteLine($"{Name} проводит совещание с командой отдела {Department}");
    }

    public void AssignTask(Employee employee, string task)
    {
        Console.WriteLine($"{Name} назначает задачу '{task}' сотруднику {employee.Name}");
    }
    
    public void ConductMeeting(string meetingTopic)
    {
        Console.WriteLine($"{Name} проводит совещание на тему: '{meetingTopic}'");
    }

    public void JoinProject(string projectName)
    {
        currentProjects.Add(projectName);
        Console.WriteLine($"{Name} присоединился к проекту '{projectName}' как менеджер");
    }
    
    public void LeaveProject(string projectName)
    {
        if (currentProjects.Remove(projectName))
            Console.WriteLine($"{Name} покинул проект '{projectName}'");
    }
}


public class Developer : Employee, IProjectMember
{
    private string specialization;
    private List<string> currentProjects = new List<string>();

    private string programmingLanguage;

    public string Specialization
    {
        get { return specialization; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                specialization = value;
            else
                Console.WriteLine("Специализация не может быть пустой!");
        }
    }
    
    public string ProgrammingLanguage
    {
        get { return programmingLanguage; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                programmingLanguage = value;
            else
                Console.WriteLine("Язык программирования не может быть пустым!");
        }
    }

    public Developer(string name, int age, decimal salary, string email, string education, int experience, string phoneNumber, string department, string specialization, string programmingLanguage) 
        : base(name, age, salary, email, education, experience, phoneNumber, department)
    {
        Specialization = specialization;
        ProgrammingLanguage = programmingLanguage;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Специализация: {Specialization}, Язык: {ProgrammingLanguage}";
    }
    
    public override string GetDetailedInfo()
    {
        return base.GetDetailedInfo() + $", Специализация: {Specialization}, Язык: {ProgrammingLanguage}, Проектов: {currentProjects.Count}";
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"{Name} берет отпуск. Согласуйте с дедлайнами проекта!");
    }

    public override void TakeVacation(int days)
    {
        if (days > 14)
            Console.WriteLine($"{Name} запрашивает длительный отпуск на {days} дней. Требуется согласование с менеджером!");
        else
            Console.WriteLine($"{Name} берет отпуск на {days} дней. Согласуйте с дедлайнами проекта!");
    }

    public override void InteractWith(Employee otherEmployee)
    {
        if (otherEmployee is Manager)
            Console.WriteLine($"{Name} получает задачи от менеджера {otherEmployee.Name}");
        else if (otherEmployee is Designer)
            Console.WriteLine($"{Name} обсуждает техническую реализацию с дизайнером {otherEmployee.Name}");
        else
            Console.WriteLine($"{Name} помогает коллеге {otherEmployee.Name}");
    }

    public void WriteCode()
    {
        Console.WriteLine($"{Name} пишет код на {ProgrammingLanguage}");
    }

    public void DebugCode()
    {
        Console.WriteLine($"{Name} отлаживает код на {ProgrammingLanguage}");
    }
    
    public void AttendTraining(string trainingTopic)
    {
        Console.WriteLine($"{Name} посещает тренинг по теме: '{trainingTopic}'");
    }

    public void JoinProject(string projectName)
    {
        currentProjects.Add(projectName);
        Console.WriteLine($"{Name} присоединился к проекту '{projectName}' как разработчик");
    }
    
    public void LeaveProject(string projectName)
    {
        if (currentProjects.Remove(projectName))
            Console.WriteLine($"{Name} покинул проект '{projectName}'");
    }
}

public class IDeveloper : Developer
{
    public bool OS { get; set; }

    public string PreferredIDE { get; set; }

    public IDeveloper(string name, int age, decimal salary, string email, string education, int experience, string phoneNumber, string department, string specialization, string programmingLanguage, bool os, string preferredIDE) 
        : base(name, age, salary, email, education, experience, phoneNumber, department, specialization, programmingLanguage)
    {
        OS = os;
        PreferredIDE = preferredIDE;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Работает на Windows: {OS}, Предпочитаемая IDE: {PreferredIDE}";
    }
    
    public override string GetDetailedInfo()
    {
        return base.GetDetailedInfo() + $", Работает на Windows: {OS}, Предпочитаемая IDE: {PreferredIDE}";
    }

    public override void UpdateContactInfo(string newEmail, string newPhone)
    {
        Email = newEmail;
        PhoneNumber = newPhone;
        Console.WriteLine($"{Name}: Контактная информация обновлена (специальная процедура для IDeveloper)");
    }
}

public class Designer : Employee, IDesigner, IProjectMember
{
    private string designStyle;
    private List<string> currentProjects = new List<string>();

    private string designTool;

    public string DesignStyle
    {
        get { return designStyle; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                designStyle = value;
            else
                Console.WriteLine("Стиль дизайна не может быть пустым!");
        }
    }
    
    public string DesignTool
    {
        get { return designTool; }
        set 
        { 
            if (!string.IsNullOrEmpty(value))
                designTool = value;
            else
                Console.WriteLine("Инструмент дизайна не может быть пустым!");
        }
    }

    public Designer(string name, int age, decimal salary, string email, string education, int experience, string phoneNumber, string department, string designStyle, string designTool) 
        : base(name, age, salary, email, education, experience, phoneNumber, department)
    {
        DesignStyle = designStyle;
        DesignTool = designTool;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Стиль: {DesignStyle}, Инструмент: {DesignTool}";
    }
    
    public override string GetDetailedInfo()
    {
        return base.GetDetailedInfo() + $", Стиль: {DesignStyle}, Инструмент: {DesignTool}, Проектов: {currentProjects.Count}";
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} создает дизайн в стиле {DesignStyle} используя {DesignTool}");
    }

    public override void TakeVacation(int days)
    {
        Console.WriteLine($"{Name} берет творческий отпуск на {days} дней для поиска вдохновения");
    }

    public override void InteractWith(Employee otherEmployee)
    {
        if (otherEmployee is Manager)
            Console.WriteLine($"{Name} получает тех задание от менеджера {otherEmployee.Name}");
        else if (otherEmployee is Developer)
            Console.WriteLine($"{Name} передает макет разработчику {otherEmployee.Name}");
        else
            Console.WriteLine($"{Name} консультирует {otherEmployee.Name} по дизайну");
    }

    public void CreatePrototype()
    {
        Console.WriteLine($"{Name} создает прототип в стиле {DesignStyle} используя {DesignTool}");
    }

    public void CreateMockup()
    {
        Console.WriteLine($"{Name} создает мокап в стиле {DesignStyle}");
    }
    
    public void PresentDesign()
    {
        Console.WriteLine($"{Name} представляет дизайн команде отдела {Department}");
    }

    public void JoinProject(string projectName)
    {
        currentProjects.Add(projectName);
        Console.WriteLine($"{Name} присоединился к проекту '{projectName}' как дизайнер");
    }
    
    public void LeaveProject(string projectName)
    {
        if (currentProjects.Remove(projectName))
            Console.WriteLine($"{Name} покинул проект '{projectName}'");
    }
}


Manager manager = new Manager("Иван Иванов", 35, 80000, "ivanov@mail.ru", "Бакалавриат", 7, "+7-900-123-45-67", "Управление", 5);
Developer developer = new Developer("Анна Петрова", 28, 70000, "petrova@mail.ru", "Магистратура", 4, "+7-900-123-45-68", "Разработка", "Backend", "C#");
IDeveloper ideveloper = new IDeveloper("Елена Кузнецова", 30, 100000, "kyznecova@mail.ru", "Высшее", 6, "+7-900-123-45-69", "Разработка", "FullStack", "Python", true, "PyCharm");
Designer designer = new Designer("Мария Попова", 25, 65000, "popova@mail.ru", "Среднее профессиональное", 3, "+7-900-123-45-70", "Дизайн", "Минимализм", "Figma");

Employee[] employees = { manager, developer, ideveloper, designer };

Console.WriteLine("ДЕТАЛЬНАЯ ИНФОРМАЦИЯ О ВСЕХ СОТРУДНИКАХ:");
foreach (var emp in employees)
{
    Console.WriteLine(emp.GetDetailedInfo());
}
Console.WriteLine();

Console.WriteLine("РАБОЧИЕ ОБЯЗАННОСТИ:");
foreach (var emp in employees)
{
    emp.Work();
}
Console.WriteLine();

Console.WriteLine("ОТПУСКНЫЕ:");
foreach (var emp in employees)
{
    emp.TakeVacation(10); 
}
Console.WriteLine();

Console.WriteLine("ВЗАИМОДЕЙСТВИЕ:");
manager.InteractWith(developer);
manager.InteractWith(developer, "новые требования к API"); 
developer.InteractWith(designer);
ideveloper.InteractWith(developer);
designer.InteractWith(manager);
Console.WriteLine();

Console.WriteLine("УНИКАЛЬНЫЕ МЕТОДЫ СОТРУДНИКОВ:");
manager.ConductMeeting();
manager.ConductMeeting("планирование нового проекта"); 
manager.AssignTask(developer, "рефакторинг модуля авторизации");
developer.WriteCode();
developer.DebugCode();
developer.AttendTraining("микросервисная архитектура");
ideveloper.WriteCode();
designer.CreatePrototype();
designer.CreateMockup();
designer.PresentDesign();
Console.WriteLine();

Console.WriteLine("РАБОТА С ИНТЕРФЕЙСОМ IProjectMember:");
IProjectMember[] projectMembers = { manager, developer, ideveloper, designer };
foreach (var member in projectMembers)
{
    member.JoinProject("Мобильное приложение");
}
Console.WriteLine();

Console.WriteLine("РАБОТА С GENERIC КЛАССОМ:");
EmployeeManager<Developer> devManager = new EmployeeManager<Developer>();
devManager.AddEmployee(developer);
devManager.AddEmployee(ideveloper as Developer);
devManager.DisplayAllEmployees(true); 
        
EmployeeManager<Employee> allEmployeesManager = new EmployeeManager<Employee>();
foreach (var emp in employees)
{
    allEmployeesManager.AddEmployee(emp);
}

```
